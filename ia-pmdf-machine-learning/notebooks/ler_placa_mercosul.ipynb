```python
import easyocr
import cv2
import matplotlib.pyplot as plt

reader = easyocr.Reader(['en', 'pt'], gpu=True)

def ler_placa_mercosul(imagem_retificada):
    resultado = reader.readtext(imagem_retificada, detail=0, allowlist='ABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789')

    if resultado:
        placa_texto = "".join(resultado).replace(" ", "")
        print(f"[OCR ORTZION] Placa Identificada: {placa_texto}")
        return placa_texto
    return None
```
