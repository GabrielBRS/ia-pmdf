```python
import faiss
import numpy as np

def executar_busca_disruptiva(embedding_capturado, base_gold_embeddings):
    d = 512

    index = faiss.IndexFlatIP(d)

    faiss.normalize_L2(base_gold_embeddings)
    index.add(base_gold_embeddings)

    faiss.normalize_L2(embedding_capturado.reshape(1, -1))
    distancias, indices = index.search(embedding_capturado.reshape(1, -1), k=1)

    score_similaridade = distancias[0][0]
    id_suspeito = indices[0][0]

    if score_similaridade > 0.88:
        print(f"[ALERTA PMDF] Suspeito ID {id_suspeito} identificado!")
        print(f"Confiança Forense: {score_similaridade * 100:.2f}%")
        return id_suspeito, score_similaridade

    return None, score_similaridade
```
