```python
import pandas as pd
import datetime

def ingest_lpr_to_bronze(source_path, bucket_bronze):

    df = pd.read_parquet(source_path)

    df['ingested_at'] = datetime.datetime.now()
    df['source_system'] = 'PMDF_RADAR_LPR'

    output_path = f"{bucket_bronze}/pmdf/lpr/{datetime.date.today()}/raw_data.parquet"
    df.to_parquet(output_path, index=False)
    print("Dados LPR movidos para Bronze.")
```

```python
def ingest_criminal_db_to_bronze(connection_string, bucket_bronze):

    query = "SELECT * FROM db_criminal.ocorrencias_veiculos"

    df = pd.read_sql(query, connection_string)

    output_path = f"{bucket_bronze}/sinesp/veiculos_roubados/{datetime.date.today()}/data.parquet"
    df.to_parquet(output_path, index=False)
    print("Base criminal integrada à Bronze.")
```
