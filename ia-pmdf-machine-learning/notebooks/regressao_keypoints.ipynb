```python
import torch
import torch.nn as nn

class OrtzionKeypointHead(nn.Module):
    def __init__(self, input_dim):
        super(OrtzionKeypointHead, self).__init__()
        self.keypoint_regressor = nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.ReLU(),
            nn.Linear(256, 8),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.keypoint_regressor(x)
```
