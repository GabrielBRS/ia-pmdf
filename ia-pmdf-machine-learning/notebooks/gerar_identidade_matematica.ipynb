```python
import numpy as np
import torch

def gerar_identidade_matematica(face_tensor, model_extractor):

    with torch.no_grad():
        embedding = model_extractor(face_tensor.unsqueeze(0))

    embedding_norm = torch.nn.functional.normalize(embedding, p=2, dim=1)

    identidade_vetorial = embedding_norm.squeeze().cpu().numpy()

    print(f"[ORTZION] Vetor Gerado. Dimensões: {identidade_vetorial.shape[0]}")
    return identidade_vetorial
```
