```python
import faiss
import numpy as np
import time

dimension = 512
n_records = 1_000_000

np.random.seed(42)
criminal_database = np.random.random((n_records, dimension)).astype('float32')
faiss.normalize_L2(criminal_database)

index = faiss.IndexFlatIP(dimension)
index.add(criminal_database)

def buscar_suspeito_rodovia(embedding_capturado):
    query_vector = embedding_capturado.reshape(1, -1).astype('float32')
    faiss.normalize_L2(query_vector)

    start_time = time.time()

    similarities, indices = index.search(query_vector, k=1)

    end_time = time.time()
    latency_ms = (end_time - start_time) * 1000

    return indices[0][0], similarities[0][0], latency_ms

suspect_face = np.random.random((dimension,)).astype('float32')
match_id, score, latency = buscar_suspeito_rodovia(suspect_face)

print(f"[SISTEMA ORTZION] Alerta Gerado em {latency:.2f} ms")
print(f"ID do Suspeito Identificado: {match_id}")
print(f"Confiança da Similaridade: {score:.4f}")
```
