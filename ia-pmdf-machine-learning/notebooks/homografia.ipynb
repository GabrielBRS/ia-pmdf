```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def retificar_placa(image, pontos_origem):
    largura, altura = 400, 130

    pontos_destino = np.array([
        [0, 0],
        [largura - 1, 0],
        [largura - 1, altura - 1],
        [0, altura - 1]
    ], dtype="float32")

    pts_src = np.array(pontos_origem, dtype="float32")

    M = cv2.getPerspectiveTransform(pts_src, pontos_destino)

    placa_retificada = cv2.warpPerspective(image, M, (largura, altura))

    return placa_retificada
```
