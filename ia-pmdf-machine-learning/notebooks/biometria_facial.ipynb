```python
import torch
import torch.nn as nn
from facenet_pytorch import InceptionResnetV1, MTCNN
import cv2
from PIL import Image

detector = MTCNN()
extractor = InceptionResnetV1(pretrained='vggface2').eval()

def extrair_embedding_facial(image_path):
    img = Image.open(image_path)
    face_img = detector(img)

    if face_img is not None:
        with torch.no_grad():
            embedding = extractor(face_img.unsqueeze(0))

        embedding_vector = embedding.squeeze().cpu().numpy()
        print(f"[ORTZION BIOMETRIA] Vetor de {len(embedding_vector)} dimensões gerado.")
        return embedding_vector

    return None
```
